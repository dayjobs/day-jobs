```python
#Example of if/else statements
#GPA conversions
#90-100 => 4.0
#80-89 => 3.0
#70-79 => 2.0
#0-69 => 0.0

average = int(input("Enter your average grade: "))


if average >= 90:
  gpa = 4.0
elif average >= 80:
  gpa = 3.0
elif average >= 70:
  gpa = 2.0
else:
  gpa = 0.0
print(f"Average: {average} = GPA: {gpa}")
```

```python
# Loop examples

number = int(input("Please enter a number: "))

while number % 2 == 0:
  number = int(input("Please enter an odd number: "))
print(f"Thank you! You inputed {number}, which is an odd number!")
```

```python
#Find the biggest number in the list
numbers = [1, 4, 5, 0, -200, 49203, 1923904, 4]
max_num = float('-inf') #define initial max number  to negative infinity
#any number in the numbers list will be greater than this initial value

for number in numbers:
  if number > max_num:
    max_num = number
    print(f"Updated max number to {max_num}")
print(f"The maxium number in the list is {max_num}")


```
